## 1️⃣ range() 함수

range() 함수는 **연속된 숫자 범위를 생성**하는 내장 함수이다.

```python
range(stop)               # 0 ~ stop-1
range(start, stop)        # start ~ stop-1
range(start, stop, step)  # step 간격으로 증가
```


````
print("range(5):")
for i in range(5):
    print(i)
````

````
print("range(2, 8):")
for i in range(2, 8):
    print(i, end=" ")
print("\n")

print("range(1, 10, 2):")
for i in range(1, 10, 2):
    print(i, end=" ")
print("\n")

print("range(10, 0, -1):")
for i in range(10, 0, -1):
    print(i, end=" ")
print()
````

## 2️⃣ enumerate() 함수

리스트를 순회할 때 **인덱스와 값**을 함께 사용하고 싶을 때가 많다.  
이때 `enumerate()`를 사용하면 코드가 간결해진다.

```python
for idx, value in enumerate(리스트, start=시작번호):
    ...
```

````
fruits = ["사과", "배", "포도"]

for idx, fruit in enumerate(fruits):
    print(idx, fruit)

print("--- start=1 옵션 사용 ---")
for idx, fruit in enumerate(fruits, start=1):
    print(idx, fruit)
````

## 3️⃣ break와 continue

| 명령어 | 설명 |
|:--|:--|
| break | 반복문을 즉시 종료한다. |
| continue | 현재 반복을 건너뛰고 다음 반복으로 진행한다. |

예시:


````
for i in range(1, 10):
    if i % 2 == 0:
        continue
    if i > 7:
        break
    print(i)
````

`break`는 **자신이 속한 가장 안쪽 반복문만** 종료한다.


````
for i in range(3):
    for j in range(3):
        if j == 1:
            break
        print("i:", i, "j:", j)
````

## 4️⃣ 리스트 컴프리헨션

리스트 컴프리헨션은 **반복문과 조건문을 이용해 리스트를 간결하게 만드는 문법**이다.

```python
[표현식 for 변수 in 반복대상 if 조건식]
```

예시:


````
nums = [1, 2, 3, 4, 5]
squares = [x**2 for x in nums]
print(squares)

even_squares = [x**2 for x in nums if x % 2 == 0]
print(even_squares)
````

## 📘 정리

| 개념 | 설명 |
|:--|:--|
| range(start, stop, step) | 다양한 간격과 방향으로 정수 시퀀스를 생성한다. |
| enumerate() | 인덱스와 값을 동시에 꺼내며 반복할 수 있다. |
| break | 현재 반복문을 즉시 종료한다. |
| continue | 현재 반복만 건너뛰고 다음 반복으로 이동한다. |
| 리스트 컴프리헨션 | 리스트를 간결하게 생성하는 문법이다. |


## 🧠 미니 실습 : 인덱스-점수 출력

### 문제

다음 요구사항을 만족하는 코드를 작성하라.

1. 점수 리스트 `scores = [90, 30, 0, 70, 80, 0, 100]`가 있다.  
2. `enumerate()`를 사용하여 **인덱스와 점수**를 함께 순회한다.  
3. 점수가 0이면 `"0점 발견, 이후 점수는 무시한다."`를 출력하고 반복을 `break`로 종료한다.  
4. 0이 나오기 전까지의 점수 중에서 **60점 이상인 점수만** 출력한다.  
   - 단, 출력 형식은 `인덱스-점수` 형태로 한다. 예) `0-90`

아래 셀에 코드를 작성해보자.

````
# 직접 코드를 작성해보자.

scores = [90, 30, 0, 70, 80, 0, 100]

# 힌트
# for idx, s in enumerate(scores):
#     ...
````

### ✅ 정답 예시

````
scores = [90, 30, 0, 70, 80, 0, 100]

for idx, s in enumerate(scores):
    if s == 0:
        print("0점 발견, 이후 점수는 무시한다.")
        break
    if s >= 60:
        print(f"{idx}-{s}")
````
