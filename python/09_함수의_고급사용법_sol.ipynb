# 함수의 고급사용법

---

## 🎯 학습 목표
이 강의에서는 함수의 **고급 사용법**을 학습한다.  
아래 개념을 배우면 함수 사용 능력이 한 단계 상승한다.

- 매개변수(Parameter) 종류
- 기본값 매개변수(Default Parameter)
- 키워드 인자(Keyword Argument)
- 가변 인자(*args)
- 가변 키워드 인자(**kwargs)
- 여러 반환값 & 언패킹

---


##✅ 매개변수(Parameter) 종류

파이썬 함수는 다음과 같은 다양한 형태의 매개변수를 지원한다.

| 종류 | 형태 | 설명 |
|------|------|------|
| 위치 인자 | `func(1, 2)` | 순서대로 전달 |
| 기본값 인자 | `def f(a, b=10)` | 기본값을 가진 인자 |
| 키워드 인자 | `func(a=1, b=2)` | 이름으로 전달 |
| *args | `def f(*args)` | 개수가 정해지지 않은 위치 인자 |
| **kwargs | `def f(**kwargs)` | key=value 형태로 여러 인자 전달 |

---


##1️⃣ 기본값 매개변수(Default Parameter)

기본값을 지정한 매개변수는 호출 시 값을 전달하지 않으면 기본값을 사용한다.


````
def greet(name, msg="안녕하세요"):
    print(name, msg)

greet("철수")
greet("영희", "반가워!")
````

##2️⃣ 키워드 인자(Keyword Arguments)

함수를 호출할 때 매개변수 이름을 함께 전달할 수 있다.

```python
def func(a, b, c=10):
    ...
```

다음과 같이 다양한 순서로 전달할 수 있다.


````
def func(a, b, c=10):
    print(a, b, c)

func(1, c=2, b=3)
````

###🔹키워드 인자 규칙

- 위치 인자는 키워드 인자보다 먼저 나와야 한다.  
- 동일한 인자를 두 번 전달하면 오류가 발생한다.


##3️⃣ 가변 인자 (*args)

여러 개수를 제한 없이 받는 매개변수이다.  
인자들은 **튜플** 형태로 전달된다.


````
def add_all(*nums):
    print(nums)
    return sum(nums)

print(add_all(1, 2, 3, 4))
````

##4️⃣ 가변 키워드 인자 (**kwargs)

`key=value` 형태로 원하는 만큼 전달할 수 있으며,  
함수 내부에서는 **딕셔너리** 형태로 관리된다.


````
def show_info(**info):
    print(info)

show_info(name="민수", age=20, city="Seoul")
````

##5️⃣ 여러 값 반환 & 언패킹

파이썬 함수는 여러 값을 반환할 수 있으며,  
반환된 값은 자동으로 **튜플**에 담긴다.


````
def get_stats(nums):
    return min(nums), max(nums), sum(nums)

low, high, total = get_stats([3, 7, 2, 10])
print(low, high, total)
````

# 🧠 미니 실습 : 프로필 업데이트

다음 요구사항을 만족하는 함수를 작성하라.

### ✔ 요구사항
1. 함수 이름은 `create_profile`로 한다.  
2. 필수 인자는 `name`, `age` 두 개이다.  
3. `**kwargs`를 사용하여 취미, 거주지 등 원하는 값을 추가로 넘길 수 있어야 한다.  
4. 모든 정보를 담은 **딕셔너리**를 반환한다.

예:
```python
create_profile("민수", 22, city="Seoul", hobby="game")
```

아래 코드 셀에 직접 작성해보자.


````
def create_profile(name, age, **kwargs):
    profile = {"name": name, "age": age}
    profile.update(kwargs)
    return profile

print(create_profile("민수", 22, city="Seoul", hobby="game"))
````
