## 1️⃣ 숫자형 (int, float)

숫자형은 **정수형(int)** 과 **실수형(float)** 으로 구분된다.

- `int`: 정수값을 저장 (예: 10, -3, 0)
- `float`: 소수점을 포함한 실수값 저장 (예: 3.14, -0.1)

### 🔹 주요 특징
- 정수형과 실수형은 함께 계산 가능하다.
- `/` 연산은 항상 **float 결과**를 반환한다.
- `//` 연산은 몫(소수점 이하 제거)을 구한다.
- `%`는 나머지, `**`는 거듭제곱을 구한다.


````python
a = 10      # int
b = 3.5     # float

print("a:", a, type(a))
print("b:", b, type(b))

print("덧셈:", a + b)
print("뺄셈:", a - b)
print("곱셈:", a * b)
print("나눗셈(/):", a / b)
print("몫(//):", a // 3)
print("나머지(%):", a % 3)
print("거듭제곱(**):", 2 ** 3)
````

## 2️⃣ 불(Boolean)

`True` 또는 `False` 두 값만 존재하는 자료형이다.  
주로 **비교 연산**이나 **조건문**의 결과로 사용된다.

### 🔹 특징
- 0, 빈 문자열(`""`), 빈 리스트(`[]`) 등은 `False`로 평가된다.
- 그 외의 값은 모두 `True`로 평가된다.


````python
x = 10
y = 7

print(x > y)      # True
print(x == y)     # False
print(x != y)     # True

print(bool(0), bool(1), bool(""), bool("Python"))
````

## 3️⃣ 문자열(String)

문자는 작은따옴표(') 또는 큰따옴표(")로 감싸서 표현한다.

```python
text = "Hello"
name = 'Python'
```

문자열은 **순서(Sequence)** 가 있기 때문에, 각 문자를 **인덱스(index)** 로 접근할 수 있고,  
일정 구간을 잘라내는 **슬라이싱(slicing)** 도 가능하다.


### 🔹 인덱싱(Indexing)

- 인덱스는 0부터 시작한다.  
- 음수 인덱스는 뒤에서부터 접근한다.

| 문자열 | P | y | t | h | o | n |
|:--:|:--:|:--:|:--:|:--:|:--:|:--:|
| 인덱스 | 0 | 1 | 2 | 3 | 4 | 5 |
| 역인덱스 | -6 | -5 | -4 | -3 | -2 | -1 |


````python
text = "Python"

print("전체:", text)
print("0번째 문자:", text[0])
print("마지막 문자:", text[-1])
````

### 🔹 슬라이싱(Slicing)

슬라이싱은 **문자열의 특정 구간을 잘라내는 기능**이다.  
형식은 `[시작:끝:간격]`으로 표현되며, **끝 인덱스는 포함되지 않는다.**

| 예시 | 의미 | 결과 |
|:--|:--|:--|
| `text[0:3]` | 0~2번째 문자 | 'Pyt' |
| `text[3:]` | 3번째부터 끝까지 | 'hon' |
| `text[:4]` | 처음부터 3번째까지 | 'Pyth' |
| `text[::2]` | 2칸 간격으로 추출 | 'Pto' |
| `text[::-1]` | 역순 | 'nohtyP' |


````python
text = "Python Programming"

print("앞 6글자:", text[:6])
print("7번째 이후:", text[7:])
print("간격 2:", text[::2])
print("역순:", text[::-1])
````

## 4️⃣ 문자열 연산 및 메서드

### 🔹 문자열 연산자
- `+` : 문자열 이어붙이기
- `*` : 문자열 반복

### 🔹 주요 문자열 메서드

| 메서드 | 설명 | 예시 |
|:--|:--|:--|
| `len(s)` | 문자열 길이 | len('hi') → 2 |
| `s.upper()` | 대문자로 변환 | 'hi'.upper() → 'HI' |
| `s.lower()` | 소문자로 변환 | 'HI'.lower() → 'hi' |
| `s.strip()` | 공백 제거 | ' hi '.strip() → 'hi' |
| `s.split(',')` | 구분자 기준으로 분리 | 'a,b'.split(',') → ['a','b'] |
| `s.replace(a,b)` | 문자열 치환 | '2025-11-11'.replace('-', '/') → '2025/11/11' |


````python
msg = "  Hello Python  "

print("이어붙이기:", "Hello" + " " + "World")
print("반복:", "Hi! " * 3)

print("길이:", len(msg))
print("strip():", msg.strip())
print("upper():", msg.upper())
print("split():", "apple,banana,grape".split(","))
print("replace():", "2025-11-11".replace("-", "/"))
````

## 5️⃣ 형 변환(Type Casting)

입력값은 기본적으로 **문자열(str)** 로 들어오기 때문에, 숫자 연산을 위해 형 변환이 필요하다.

| 함수 | 설명 | 예시 |
|:--|:--|:--|
| `int()` | 문자열 → 정수 | int("10") → 10 |
| `float()` | 문자열 → 실수 | float("3.14") → 3.14 |
| `str()` | 숫자 → 문자열 | str(10) → "10" |


````python
# 문자열 -> 숫자
num_str = "10"
num = int(num_str)
print(num, type(num))

# 숫자 -> 문자열
n = 3.14
s = str(n)
print(s, type(s))

# 문자열 -> 실수
f = float("2.718")
print(f, type(f))
````

## 6️⃣ input()과 형 변환

`input()` 함수는 항상 **문자열(str)** 로 반환된다.  
숫자 계산을 하려면 `int()` 또는 `float()`로 감싸서 변환해야 한다.

```python
age = int(input("나이를 입력하세요: "))
print(f"10년 후 나이는 {age + 10}살입니다.")
```


## 📘 정리

1. `int`, `float`는 숫자 연산의 기본 자료형이다.  
2. `bool`은 참/거짓을 표현하며 조건식 결과로 자주 사용된다.  
3. 문자열은 인덱싱, 슬라이싱을 통해 부분 추출이 가능하다.  
4. 문자열 메서드를 활용하면 데이터를 쉽게 가공할 수 있다.  
5. `input()`으로 받은 값은 문자열이므로, 필요 시 형 변환이 필요하다.
