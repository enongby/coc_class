## 🧩 제어문의 역할

제어문은 프로그램의 **실행 흐름을 제어하는 문장**이다.  
조건에 따라 코드를 선택적으로 실행하거나, 반복해서 실행할 수 있게 해주는 도구이다.

파이썬에서 대표적인 제어문은 다음과 같다.

- `if` 조건문  
- `while` 반복문  
- `for` 반복문  

## 1️⃣ if문

조건문은 특정 조건이 참(True)일 때 코드를 실행하는 제어 구조이다.  
조건이 거짓(False)이면 실행되지 않는다.

형식:
```python
if 조건식:
    실행문
elif 조건식2:
    실행문
else:
    실행문
```


````python
score = 85

if score >= 90:
    print("A학점이다.")
elif score >= 80:
    print("B학점이다.")
else:
    print("C학점 이하이다.")
````

###❗ 조건문에서 자주 발생하는 실수

1. **`=`와 `==` 혼동**  
   - `=` : 대입 연산자 (값을 넣을 때 사용)  
   - `==` : 비교 연산자 (같은지 비교할 때 사용)

2. **들여쓰기 불일치**  
   - 같은 블록에 속하는 코드는 들여쓰기를 동일하게 맞춰야 한다.

3. **`if x:` 의미 오해**  
   - 0, 빈 문자열, 빈 리스트는 거짓으로 취급된다.  
   - 그 외 값은 참으로 취급된다.

## 2️⃣ 비교 및 논리 연산자

조건식에는 비교 연산자와 논리 연산자를 함께 사용할 수 있다.

| 구분 | 연산자 | 의미 |
|:--|:--|:--|
| 비교 연산자 | == | 같다 |
| 비교 연산자 | != | 다르다 |
| 비교 연산자 | >, <, >=, <= | 크다/작다 비교 |
| 논리 연산자 | and | 두 조건이 모두 참 |
| 논리 연산자 | or | 두 조건 중 하나라도 참 |
| 논리 연산자 | not | 조건의 반대 |

예시:


````python
x = 10
y = 5

if x > y and x < 20:
    print("x는 y보다 크고 20보다 작다.")
````

## 3️⃣ 중첩 if문

if문 안에 또 다른 if문이 들어갈 수 있다.  
조건 안에서 다시 세부 조건을 검사할 때 사용한다.

````python
age = 20
has_ticket = True

if age >= 19:
    if has_ticket:
        print("입장 가능하다.")
    else:
        print("티켓이 필요하다.")
else:
    print("미성년자는 입장할 수 없다.")
````

## 4️⃣ while문

while문은 조건이 참(True)인 동안 반복을 수행한다.  
조건이 거짓(False)이 되면 반복이 종료된다.


````python
# 1부터 5까지 출력하기
i = 1

while i <= 5:
    print(i)
    i += 1
````

## 5️⃣ for문

for문은 정해진 횟수만큼 반복을 수행한다.

```python
for 변수 in 반복대상:
    실행문
```

가장 많이 사용하는 형태는 `range()`와 함께 사용하는 방식이다.

```python
for i in range(5):
    print(i)
```

`range(5)`는 0부터 4까지의 정수를 생성한다.


````python
print("0부터 4까지 출력")
for i in range(5):
    print(i)
````

for문은 리스트와 문자열에도 그대로 사용할 수 있다.

````python
fruits = ["사과", "배", "포도"]

for f in fruits:
    print("과일:", f)

text = "Python"
for ch in text:
    print("문자:", ch)
````

## 6️⃣ 중첩 반복문

반복문 안에 또 다른 반복문을 작성할 수 있다.  
2차원 데이터(표, 행렬 등)를 처리할 때 유용하다.


````python
for i in range(4):
    for j in range(4):
        print(i, j)
````

## 📘 정리

| 구분 | 형태 | 설명 |
|:--|:--|:--|
| if | `if 조건:` | 조건이 참일 때만 코드 실행 |
| if-else | `if ~ else` | 조건에 따라 둘 중 하나 실행 |
| if-elif-else | `if ~ elif ~ else` | 여러 조건을 순서대로 검사 |
| while | `while 조건:` | 조건이 참인 동안 반복 |
| for | `for 변수 in 객체:` | 객체의 요소를 하나씩 꺼내며 반복 |


## 🧠 미니 실습: 합격·불합격 판정 프로그램

1. 세 과목의 점수를 입력받는다.  
2. 각 과목이 모두 40점 이상이고, 세 과목의 평균이 60점 이상이면 `"합격이다."`  
3. 그 외의 경우 `"불합격이다."` 출력한다.


````python
total = 0
is_pass_subjects = True # 과목별 통과 여부 확인 변수

for i in range(3):
    score = int(input(f"{i+1}번째 과목 점수를 입력하세요: "))

    if score < 40:
        is_pass_subjects = False  # 하나라도 40점 미만이면 False로 설정

    total += score

avg = total / 3

if is_pass_subjects and avg >= 60:
    print("합격이다.")
else:
    print("불합격이다.")
````
