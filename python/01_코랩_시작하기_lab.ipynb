## 1️⃣ Google Colab이란?

Google Colab(코랩)은 Google에서 제공하는 **클라우드 기반 파이썬 실행 환경**이다.

- 설치 없이 **브라우저에서 바로 파이썬 실행**
- 노트북(.ipynb) 형태의 문서를 셀 단위로 실행
- GPU/TPU 제공(무료 버전도 가능)
- Google Drive와 연동되어 파일 저장 및 업로드가 가능
- 파이썬 인터프리터 기반으로 코드를 한 줄씩 실행하고 즉시 결과 확인 가능

즉, **코랩은 Google이 서버를 제공하고 우리는 브라우저로 접속해 파이썬을 실행하는 형태의 개발 환경**이다.

## 2️⃣ 코랩에서의 '셀(Cell)' 구조

코랩은 코드를 **셀 단위로** 나눠서 실행한다.

예시 코드:
```python
a = 1
print(a)
```

셀을 실행하면 바로 아래에 결과가 출력되며,  
코드를 조금씩 테스트하고 수정하기 좋다.

## 3️⃣ 코랩에서 자주 사용하는 단축키

### ▶ 실행 관련
| 단축키 | 기능 |
|-------|------|
| **Shift + Enter** | 현재 셀 실행 후 다음 셀로 이동 |
| **Ctrl + Enter** | 현재 셀만 실행 |
| **Alt + Enter** | 현재 셀 실행 후 아래 새 셀 생성 |

### ▶ 셀 추가 & 삭제
| 단축키 | 기능 |
|-------|------|
| **Ctrl + M, A** | 위에 새 셀 추가 |
| **Ctrl + M, B** | 아래에 새 셀 추가 |
| **Ctrl + M, D** | 현재 셀 삭제 |

### ▶ 셀 복사/붙여넣기/자르기
| 단축키 | 기능 |
|-------|------|
| **Ctrl + M, X** | 셀 자르기 |
| **Ctrl + M, C** | 셀 복사 |
| **Ctrl + M, V** | 셀 붙여넣기 |

### ▶ 셀 타입 변경
| 단축키 | 기능 |
|-------|------|
| **Ctrl + M, Y** | 코드 셀로 변경 |
| **Ctrl + M, M** | 마크다운 셀로 변경 |

### ▶ 편집 관련
| 단축키 | 기능 |
|-------|------|
| **Ctrl + /** | 현재 줄 주석 처리/해제 |
| **Ctrl + F** | 문서 내 검색 |
| **Ctrl + Shift + F** | 전체 노트북에서 검색 |
