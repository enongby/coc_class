## 1️⃣ 리스트(List)

리스트는 여러 데이터를 순서대로 저장하는 자료형이다.  
- 대괄호 `[]`로 표현한다.  
- 인덱스를 이용해 개별 요소에 접근할 수 있다.  
- 값을 추가하거나 수정할 수 있는 **mutable(가변형)** 자료형이다.


````
fruits = ["사과", "바나나", "포도"]
print("리스트:", fruits)
print("첫 번째:", fruits[0])
print("마지막:", fruits[-1])

# 요소 추가 및 삭제
fruits.append("복숭아")
print("추가 후:", fruits)
fruits.remove("바나나")
print("삭제 후:", fruits)

# 인덱싱과 슬라이싱
print("앞 2개:", fruits[:2])
````

### ✅ 리스트 주요 메서드

| 메서드 | 설명 | 예시 |
|:--|:--|:--|
| append(x) | 맨 뒤에 추가 | fruits.append("체리") |
| insert(i, x) | 위치 지정 추가 | fruits.insert(1, "키위") |
| remove(x) | 값 삭제 | fruits.remove("사과") |
| pop(i) | i번째 값 꺼내기 | fruits.pop(0) |
| sort() | 정렬 | fruits.sort() |
| reverse() | 역순 정렬 | fruits.reverse() |

## 2️⃣ 튜플(Tuple)

튜플은 리스트와 유사하지만 **수정이 불가능(immutable)** 하다.  
- 소괄호 `()`로 표현한다.  
- 데이터가 변경되지 않아야 하는 경우에 사용한다.  
- 리스트보다 메모리 효율이 높고 속도가 빠르다.


````
colors = ("빨강", "초록", "파랑")
print("튜플:", colors)
print("두 번째 요소:", colors[1])

# colors[1] = "노랑"  # 오류 발생: TypeError
````

## 3️⃣ 딕셔너리(Dictionary)

딕셔너리는 **키(key)** 와 **값(value)** 쌍으로 데이터를 저장하는 자료형이다.  
- 중괄호 `{}`로 표현한다.  
- 키는 중복될 수 없다.  
- 빠르게 특정 값을 찾을 때 유용하다.


````
person = {"이름": "홍길동", "나이": 25, "직업": "개발자"}
print("딕셔너리:", person)
print("이름:", person["이름"])

# 값 수정 및 추가
person["나이"] = 26
person["취미"] = "등산"
print("수정 후:", person)

# 주요 메서드
print("keys():", person.keys())
print("values():", person.values())
print("items():", person.items())
````

## 4️⃣ 세트(Set)

세트는 **중복을 허용하지 않고, 순서가 없는** 자료형이다.  
- 중괄호 `{}`로 표현한다.  
- 집합 연산(합집합, 교집합, 차집합 등)을 수행할 수 있다.


````
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("합집합:", A | B)
print("교집합:", A & B)
print("차집합:", A - B)

A.add(7)
print("요소 추가 후:", A)
````

## 5️⃣ 컬렉션 간 형 변환

리스트, 튜플, 세트는 서로 변환할 수 있다.

```python
list()  → 리스트로 변환  
tuple() → 튜플로 변환  
set()   → 세트로 변환  
```

중복된 데이터를 제거하거나 형태를 바꿀 때 유용하다.


````
nums = [1, 2, 2, 3, 3, 4]
print("원본 리스트:", nums)

unique = set(nums)
print("세트(중복 제거):", unique)

again_list = list(unique)
print("리스트로 변환:", again_list)
````

## 📘 정리

| 자료형 | 특징 | 기호 |
|:--|:--|:--|
| 리스트 | 순서 O, 수정 가능 | [] |
| 튜플 | 순서 O, 수정 불가 | () |
| 딕셔너리 | 키-값 구조 | {} |
| 세트 | 순서 X, 중복 X | {} |

각 컬렉션은 상황에 맞게 선택하여 사용한다.


## 🧠 미니 실습: 연락처 관리 프로그램

1. 두 명의 사용자에게 이름과 전화번호를 각각 입력받는다.  
2. 두 사람의 정보를 딕셔너리에 저장한다.  
3. 저장된 정보를 출력한다.


````
contacts = {}

# 첫 번째 사람
name1 = input("이름을 입력한다: ")
phone1 = input("전화번호를 입력한다: ")
contacts[name1] = phone1

# 두 번째 사람
name2 = input("이름을 입력한다: ")
phone2 = input("전화번호를 입력한다: ")
contacts[name2] = phone2

# 출력
print("=== 등록된 연락처 ===")
print(name1, ":", contacts[name1])
print(name2, ":", contacts[name2])
````
