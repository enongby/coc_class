## 1️⃣ 함수의 개념

함수는 프로그램에서 **반복되는 코드를 하나로 묶어 재사용할 수 있게 만든 구조**이다.  
복잡한 프로그램을 구조화하고 유지보수를 쉽게 만든다.

```python
def 함수이름(매개변수):
    실행문
    return 결과
```


````
def greet():
    print("안녕하세요!")

greet()
````

## 2️⃣ 매개변수와 반환값

함수는 입력값(매개변수)을 받아 연산 후 결과를 반환(return)할 수 있다.


````
def add(a, b):
    return a + b

result = add(3, 5)
print("결과:", result)
````

## 3️⃣ 기본값 매개변수

매개변수에 기본값을 설정하면, 인자를 생략해도 함수를 호출할 수 있다.


````
def introduce(name, age=20):
    print(f"이름: {name}, 나이: {age}")

introduce("홍길동")
introduce("이몽룡", 25)
````

## 4️⃣ 지역 변수와 전역 변수

- **지역 변수(Local Variable)** : 함수 내부에서만 사용되는 변수  
- **전역 변수(Global Variable)** : 프로그램 전체에서 사용 가능한 변수


````
x = 10

def show():
    x = 5
    print("함수 내부:", x)

show()
print("함수 외부:", x)
````

## 5️⃣ 모듈(Module)의 개념

모듈은 함수, 변수, 클래스를 하나의 파일(.py)로 묶은 것이다.  
다른 파이썬 파일에서 불러와 사용할 수 있다.

```python
import math

print(math.sqrt(16))
print(math.pi)
```

모듈을 사용하면 코드 재사용성과 유지보수성이 높아진다.


## 6️⃣ import의 다양한 방식

| 형식 | 설명 | 예시 |
|:--|:--|:--|
| `import 모듈` | 전체 모듈을 불러온다. | `import math` |
| `import 모듈 as 별칭` | 모듈 이름이 길 때 별칭으로 사용한다. | `import math as m` |
| `from 모듈 import 함수` | 특정 함수만 불러온다. | `from math import sqrt` |
| `from 모듈 import *` | 모든 구성요소를 불러온다. | `from math import *` |

예시:


````
from math import sqrt, pi

print("제곱근:", sqrt(25))
print("원주율:", pi)
````

## 7️⃣ 외부 모듈 소개 (NumPy, Pandas 등)

표준 모듈 외에도 다양한 **외부 모듈(external library)** 이 존재한다.  
데이터 분석, 수치 계산, 시각화 등에 활용된다.  
`pip install` 명령어를 이용해 설치 후 `import`하여 사용할 수 있다.

| 모듈 | 주요 기능 | 예시 코드 |
|:--|:--|:--|
| numpy | 수치 계산, 배열 연산 | `import numpy as np` |
| pandas | 데이터프레임(표 형식 데이터) 처리 | `import pandas as pd` |
| matplotlib | 시각화 (그래프 그리기) | `import matplotlib.pyplot as plt` |

예시:


````
import numpy as np
arr = np.array([1, 2, 3])
print("NumPy 배열:", arr)

import pandas as pd
data = {"이름": ["홍길동", "이몽룡"], "나이": [25, 28]}
df = pd.DataFrame(data)
print(df)
````


## 📌 자주 사용하는 파이썬 내장함수 정리표

### 🔹 기본 자료형 처리 관련 내장함수

| 함수 | 설명 | 예시 | 결과 |
|------|------|------|------|
| `len(x)` | 길이(요소 개수)를 반환한다. | `len("Python")` | `6` |
| `type(x)` | 자료형을 알려준다. | `type(3.14)` | `<class 'float'>` |
| `int(x)` | 정수형으로 변환한다. | `int("10")` | `10` |
| `float(x)` | 실수형으로 변환한다. | `float("3.14")` | `3.14` |
| `str(x)` | 문자열로 변환한다. | `str(100)` | `"100"` |

---

### 🔹 수치 연산 관련 내장함수

| 함수 | 설명 | 예시 | 결과 |
|------|------|------|------|
| `max(iterable)` | 가장 큰 값을 반환한다. | `max([1,7,3])` | `7` |
| `min(iterable)` | 가장 작은 값을 반환한다. | `min([1,7,3])` | `1` |
| `sum(iterable)` | 합계를 반환한다. | `sum([1,2,3])` | `6` |
| `abs(x)` | 절댓값을 반환한다. | `abs(-10)` | `10` |
| `round(x, n)` | 반올림한다. | `round(3.14159, 2)` | `3.14` |

---

### 🔹 반복 처리 관련 내장함수

| 함수 | 설명 | 예시 | 결과 |
|------|------|------|------|
| `range()` | 정수 시퀀스 생성 | `list(range(3))` | `[0,1,2]` |
| `enumerate()` | 인덱스 + 값을 함께 반환 | `list(enumerate(["a","b"]))` | `[(0,'a'), (1,'b')]` |
