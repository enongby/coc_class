##1️⃣ 노트북 환경의 실행 구조 이해하기

코랩(Colab)과 주피터 노트북은 **문서의 위/아래와 관계없이, 실행한 순서대로 메모리에 값이 저장되는 구조**이다.

예시:

셀 1:
```python
a = 5
```

셀 2:
```python
c = a + 3
```

셀 3:
```python
a = 3
```

### 실행 순서에 따라 결과가 달라진다

- 셀1 → 셀2 → 셀3 실행 → `c = 8`
- 셀1 → 셀3 → 셀2 실행 → `c = 6`

즉, 노트북은 **위에서 아래로가 아니라, 실행 순서가 핵심**이다.

## 3️⃣ 커널(Kernel)의 역할

커널은 파이썬을 실제로 실행하는 엔진이다.

### ✔ 커널이 저장하는 정보
- 변수
- 함수
- import된 모듈
- 모든 실행 상태

### ✔ 커널 재시작이 필요한 이유
- 메모리가 꼬였을 때
- 변수 충돌이 발생할 때
- 노트북 실행이 이상하게 동작할 때

커널 재시작 = 모든 메모리 초기화

---


## 4️⃣ 코랩에서 꼭 알아야 하는 필수 기능

### ✔ (1) 런타임 재시작
메뉴 → *런타임 → 세션 다시 시작*  
→ 모든 변수 초기화

### ✔ (2) 전체 셀 실행
메뉴 → *런타임 → 세션 다시 시작 및 모두 실행*  
→ 위에서 아래로 순서대로 새 상태로 실행

### ✔ (3) 셀 실행 번호 확인
좌측 `[1]`, `[2]`, `[15]`는 **실행 순서**를 의미한다.

## 🎯 정리

- 코랩/주피터는 **실행 순서 기반 메모리 모델**을 가진다.  
- 실행 순서가 꼬이면 예측 불가능한 오류 발생 가능.  
- 커널 재시작은 실무에서 자주 사용하는 중요한 기능.
