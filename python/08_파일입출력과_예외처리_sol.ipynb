## 1️⃣ 파일 입출력의 개념과 encoding

파일 입출력은 프로그램 외부의 데이터를 파일로 **저장(출력)**하거나,  
저장된 파일을 다시 **읽어오는(입력)** 과정을 말한다.

```python
# 파일 열기 모드
# 'r' : 읽기(read)
# 'w' : 쓰기(write, 기존 내용 삭제)
# 'a' : 추가(append)
# 't' : 텍스트 모드 (기본값)
# 'b' : 바이너리 모드
```

### 🔤 encoding의 의미

파일을 열 때는 반드시 `encoding`을 지정하는 것이 좋다.  
`encoding="utf-8"`은 한글을 포함한 다양한 문자를 올바르게 저장하고 읽기 위한 표준이다.

> 💡 인코딩(encoding)은 문자를 컴퓨터가 이해할 수 있는 **이진 코드(0과 1)** 로 변환하는 과정이다.  
> 반대로 이를 다시 문자로 변환하는 과정을 **디코딩(decoding)** 이라고 한다.


## 2️⃣ 파일 쓰기

````python
f = open("test.txt", "w", encoding="utf-8")
f.write("안녕하세요. 파이썬 파일 입출력입니다.\n")
f.write("이 줄은 두 번째 줄입니다.\n")
f.close()

print("파일 저장 완료!")
````

## 3️⃣ 파일 읽기

파일의 내용을 읽어오는 주요 메서드는 다음과 같다.

| 메서드 | 설명 |
|:--|:--|
| read() | 파일 전체 내용을 문자열로 읽는다. |
| readline() | 한 줄씩 읽는다. |
| readlines() | 모든 줄을 리스트 형태로 읽는다. |


````python
f = open("test.txt", "r", encoding="utf-8")
print("=== read() 예시 ===")
print(f.read())
f.close()
````

````python
f = open("test.txt", "r", encoding="utf-8")
print("=== readline() 예시 ===")
print(f.readline())
f.close()  
````

````python
f = open("test.txt", "r", encoding="utf-8")
print("=== readlines() 예시 ===")
lines = f.readlines()
print(lines)
f.close()
````

## 4️⃣ with open() 구문

`with open()` 구문을 사용하면 파일을 자동으로 닫을 수 있어 안전하다.

```python
with open("파일명", "모드", encoding="utf-8") as f:
    f.write("내용")
```

이 방식은 파일을 열고 나서 별도의 `close()` 명령 없이도 자동으로 닫아준다.


````python
with open("test.txt", "a", encoding="utf-8") as f:
    f.write("with open() 구문으로 추가된 줄입니다.\n")

print("파일 추가 완료!")
````

## 5️⃣ 예외 처리 (try-except)

예외(Exception)는 프로그램 실행 중 발생할 수 있는 오류를 의미한다.  
`try-except` 문을 사용하면 오류가 발생해도 프로그램이 비정상 종료되지 않는다.

```python
try:
    실행할 코드
except 예외이름:
    예외 발생 시 실행할 코드
finally:
    항상 실행되는 코드
```

- **try** : 오류가 발생할 수 있는 코드를 작성  
- **except** : 오류가 발생했을 때 실행할 코드  
- **finally** : 오류 여부와 상관없이 항상 실행되는 코드 (예: 파일 닫기)


````python
try:
    x = int(input("숫자를 입력: "))
    result = 10 / x
    print("결과:", result)
except ZeroDivisionError:
    print("0으로 나눌 수 없다.")
except ValueError:
    print("숫자만 입력할 수 있다.")
finally:
    print("프로그램 종료.")
````

## 6️⃣ 파일 입출력에서의 예외 처리

파일이 존재하지 않거나 읽기 권한이 없을 때 오류가 발생할 수 있다.  
이때 `FileNotFoundError`를 처리하여 프로그램이 정상적으로 종료되도록 한다.


````python
try:
    with open("data.txt", "r", encoding="utf-8") as f:
        print(f.read())
except FileNotFoundError:
    print("파일이 존재하지 않는다.")
````

## 📘 정리

| 구분 | 설명 |
|:--|:--|
| open() | 파일 열기 (읽기, 쓰기, 추가) |
| close() | 파일 닫기 |
| with open() | 자동으로 닫힘 처리 |
| read() / write() | 파일 읽기 / 쓰기 |
| encoding | 문자 데이터를 인코딩/디코딩할 때 사용 |
| try-except | 오류 발생 시 프로그램 비정상 종료 방지 |
| finally | 항상 실행되는 코드 블록 (정리 작업 등) |


## 🧠 미니 실습: 숫자 데이터 파일 읽기 및 평균 계산

1. `"numbers.txt"` 파일에는 여러 줄에 걸쳐 숫자가 하나씩 저장되어 있다.  
2. 이 숫자들을 모두 읽어 평균을 계산한다.  
3. 파일이 없으면 `"파일을 찾을 수 없습니다."`를 출력한다.

**예시 파일 (numbers.txt)**  
```
10
20
30
40
50
```


````python
try:
    with open("numbers.txt", "r", encoding="utf-8") as f:
        nums = f.readlines()
        nums = [int(x.strip()) for x in nums]
        avg = sum(nums) / len(nums)
    print("평균:", avg)
except FileNotFoundError:
    print("파일을 찾을 수 없습니다.")
````
