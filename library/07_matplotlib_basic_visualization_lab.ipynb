## 🔍 문제 설명
Matplotlib은 데이터를 그래프와 차트로 표현하는 Python의 대표적인 시각화 라이브러리입니다.

```python
import numpy as np
import matplotlib.pyplot as plt

months = np.array([1, 2, 3, 4, 5, 6])
sales = np.array([120, 135, 128, 160, 175, 190])

print("월:", months)
print("매출:", sales)
```

### 해답 7-1: 선 그래프

```python
plt.figure(figsize=(7, 4))
plt.plot(months, sales, marker='o')
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

### 해답 7-2: 막대그래프

```python
products = ["A", "B", "C", "D"]
quantities = [35, 50, 28, 42]

plt.figure(figsize=(7, 4))
plt.bar(products, quantities)
plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Quantity")
plt.show()
```

### 해답 7-3: 산점도

```python
study_hours = [1, 2, 3, 4, 5, 6, 7]
scores = [55, 60, 66, 72, 78, 85, 90]

plt.figure(figsize=(7, 4))
plt.scatter(study_hours, scores)
plt.title("Study Hours and Scores")
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.show()
```

## 🔧 주요 함수
- `plt.plot()`: 선 그래프
- `plt.bar()`: 막대그래프
- `plt.scatter()`: 산점도
- `plt.title()`: 그래프 제목
- `plt.xlabel()`, `plt.ylabel()`: 축 이름
- `plt.grid()`: 격자 표시
- `plt.show()`: 그래프 출력
