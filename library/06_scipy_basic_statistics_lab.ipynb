# 6. SciPy 기초 실습

## 📚 학습 목표
- SciPy를 활용한 기초 통계 분석
- 두 집단 평균 차이에 대한 가설검정
- 간단한 최적화 문제 해결

## 🔍 문제 설명
SciPy는 NumPy를 기반으로 통계, 최적화, 신호처리 등 고급 과학 계산 기능을 제공하는 라이브러리입니다.

```python
import numpy as np
from scipy import stats
from scipy.optimize import minimize_scalar

# 실습 데이터: A반과 B반의 시험 점수
class_a = np.array([78, 82, 85, 88, 90, 76, 84, 87])
class_b = np.array([72, 75, 79, 81, 77, 74, 80, 76])

print("A반 평균:", class_a.mean())
print("B반 평균:", class_b.mean())
```

### 해답 6-1: 기초 통계량 계산

```python
# SciPy를 이용한 기초 통계량 확인
result = stats.describe(class_a)

print("데이터 개수:", result.nobs)
print("최솟값 / 최댓값:", result.minmax)
print("평균:", result.mean)
print("분산:", result.variance)
```

### 해답 6-2: 두 집단 평균 비교

```python
# 독립표본 t-검정
t_stat, p_value = stats.ttest_ind(class_a, class_b)

print(f"t 통계량: {t_stat:.3f}")
print(f"p-value: {p_value:.3f}")

if p_value < 0.05:
    print("두 집단의 평균 차이가 통계적으로 유의합니다.")
else:
    print("두 집단의 평균 차이가 통계적으로 유의하지 않습니다.")
```

### 해답 6-3: 간단한 최적화

```python
# 함수 f(x) = (x - 3)^2 + 2의 최솟값 찾기
def objective(x):
    return (x - 3) ** 2 + 2

result = minimize_scalar(objective)

print(f"최적의 x값: {result.x:.3f}")
print(f"함수의 최솟값: {result.fun:.3f}")
```

## 🔧 주요 함수
- `stats.describe()`: 개수, 평균, 분산 등 요약 통계량 계산
- `stats.ttest_ind()`: 서로 독립인 두 집단의 평균 비교
- `minimize_scalar()`: 하나의 변수로 이루어진 함수의 최솟값 탐색
